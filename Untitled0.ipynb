```python
"""
DL Day 6 - Project Brief
Dataset: Chest X-Ray Pneumonia Detection

1. Which dataset did you choose and why?

I chose the Chest X-Ray Pneumonia dataset because it is a real-world
medical imaging problem with significant practical impact. It allows me
to apply deep learning techniques to healthcare data and build a model
that can assist in disease detection. I also chose it because medical
image classification is an impressive portfolio project that demonstrates
computer vision skills using convolutional neural networks.

2. What is the ML task? (binary classification, multi-class, regression?)

This is a binary classification problem. The model must classify each
chest X-ray image into one of two classes: Normal or Pneumonia.
The goal is to learn patterns in X-ray images that indicate the presence
or absence of pneumonia and predict the correct class for unseen images.

3. What is the primary metric you will use and why?

The primary metric I will use is Sensitivity (Recall). In a medical
diagnosis setting, missing a patient who actually has pneumonia can be
much more dangerous than incorrectly flagging a healthy patient. While
accuracy is useful, recall is more important because it measures how
effectively the model identifies actual pneumonia cases.

4. What does a false negative mean for your chosen task?

A false negative occurs when the model predicts that a patient is
Normal even though the patient actually has Pneumonia. This is the most
dangerous type of error because the patient may not receive timely
medical treatment. In healthcare applications, minimizing false negatives
is often more important than maximizing overall accuracy.

5. What architecture will you use and why?

I will use a Convolutional Neural Network (CNN) because the dataset
consists of images. CNNs are specifically designed for computer vision
tasks and can automatically learn spatial features such as edges,
textures, shapes, and medical abnormalities from images. They are the
standard architecture for image classification problems.

6. Will you use transfer learning? Justify.

Yes, I plan to use transfer learning with a pre-trained CNN such as
MobileNetV2, ResNet50, or EfficientNet. These models have already learned
useful visual features from millions of images and can be fine-tuned for
pneumonia detection. Transfer learning usually improves accuracy,
reduces training time, and performs better when the dataset size is
limited compared to training a CNN completely from scratch.
"""
```

```python
# ==========================================
# SECTION 3 - PREPROCESSING
# ==========================================

from tensorflow import keras
import tensorflow as tf
import matplotlib.pyplot as plt
import numpy as np

IMG_SIZE = 150
BATCH_SIZE = 32

# ------------------------------------------
# Data Augmentation
# ------------------------------------------

augmentation = keras.Sequential([
    keras.layers.RandomFlip("horizontal"),
    keras.layers.RandomRotation(0.05),
    keras.layers.RandomZoom(0.10)
], name="augmentation")

# ------------------------------------------
# Dataset Loader
# ------------------------------------------

def make_dataset(directory, augment=False):

    ds = keras.utils.image_dataset_from_directory(
        directory,
        image_size=(IMG_SIZE, IMG_SIZE),
        batch_size=BATCH_SIZE,
        label_mode="binary",
        shuffle=True
    )

    # Normalize to [0,1]
    ds = ds.map(
        lambda x, y: (x / 255.0, y),
        num_parallel_calls=tf.data.AUTOTUNE
    )

    if augment:
        ds = ds.map(
            lambda x, y:
            (augmentation(x, training=True), y),
            num_parallel_calls=tf.data.AUTOTUNE
        )

    return ds.prefetch(tf.data.AUTOTUNE)


# ------------------------------------------
# Create Datasets
# ------------------------------------------

train_ds = make_dataset(
    "/content/drive/MyDrive/chest_xray/chest_xray/train",
    augment=True
)

val_ds = make_dataset(
    "/content/drive/MyDrive/chest_xray/chest_xray/val"
)

test_ds = make_dataset(
    "/content/drive/MyDrive/chest_xray/chest_xray/test"
)

# ==========================================
# Class Names
# ==========================================

print("=" * 50)
print("CLASS INFORMATION")
print("=" * 50)

print("Class Names:")
class_names = ["NORMAL", "PNEUMONIA"]

print(class_names)

for i, name in enumerate(class_names):
    print(f"{name} -> {i}")

print("\nLabel Encoding:")


# Usually:
# NORMAL -> 0
# PNEUMONIA -> 1

# ==========================================
# Dataset Shapes
# ==========================================

print("\nDATASET SHAPES")

for images, labels in train_ds.take(1):

    print("Train Batch Shape :", images.shape)
    print("Train Label Shape :", labels.shape)

for images, labels in val_ds.take(1):

    print("Val Batch Shape   :", images.shape)
    print("Val Label Shape   :", labels.shape)

for images, labels in test_ds.take(1):

    print("Test Batch Shape  :", images.shape)
    print("Test Label Shape  :", labels.shape)

# ==========================================
# Confirm Image Size
# ==========================================

print("\nIMAGE SIZE CHECK")

for images, labels in train_ds.take(1):

    print("Expected Shape:")
    print("(batch, 150, 150, 3)")

    print("Actual Shape:")
    print(images.shape)

# ==========================================
# NaN Check
# ==========================================

print("\nNaN CHECK")

for images, labels in train_ds.take(1):

    print(
        "Images contain NaN:",
        tf.math.reduce_any(
            tf.math.is_nan(images)
        ).numpy()
    )

    print(
        "Labels contain NaN:",
        tf.math.reduce_any(
            tf.math.is_nan(labels)
        ).numpy()
    )

# ==========================================
# Visualize Augmented Images
# ==========================================

# Original Dataset
original_ds = keras.utils.image_dataset_from_directory(
    "/content/drive/MyDrive/chest_xray/chest_xray/train",
    image_size=(IMG_SIZE, IMG_SIZE),
    batch_size=5,
    shuffle=True
)

orig_images, labels = next(iter(original_ds))

aug_images = augmentation(
    orig_images,
    training=True
)

fig, axes = plt.subplots(
    2,
    5,
    figsize=(12, 6)
)

for i in range(5):

    # Original
    axes[0, i].imshow(
        orig_images[i].numpy().astype("uint8")
    )

    axes[0, i].set_title("Original")
    axes[0, i].axis("off")

    # Augmented
    axes[1, i].imshow(
        aug_images[i].numpy().astype("uint8")
    )

    axes[1, i].set_title("Augmented")
    axes[1, i].axis("off")

plt.suptitle("Original vs Augmented Images")

plt.tight_layout()

plt.savefig(
    "preprocess_augmented.png",
    dpi=150
)

plt.show()

# ==========================================
# Class Imbalance Check
# ==========================================

normal_count = 1341
pneumonia_count = 3875

total = normal_count + pneumonia_count

print("\nCLASS IMBALANCE")

print(
    f"Normal: {normal_count}"
)

print(
    f"Pneumonia: {pneumonia_count}"
)

print(
    f"Normal %: {normal_count/total:.2%}"
)

print(
    f"Pneumonia %: {pneumonia_count/total:.2%}"
)

# ==========================================
# COMMENTS FOR REPORT
# ==========================================

"""
Preprocessing Answers

1. Why do you augment only the training set?

Data augmentation is used only on the training set because
its purpose is to create additional variations of training
examples and improve generalization.

Validation and test sets must remain unchanged so they
represent real-world unseen data. If augmentation is applied
to validation or test data, evaluation results become
unreliable and no longer reflect actual model performance.

2. Why normalize pixel values to [0,1]?

Raw image pixels range from 0 to 255.

Large values can slow training and cause unstable gradients.
Dividing by 255 scales all values into the range [0,1],
which improves numerical stability and helps neural networks
converge faster.

3. How did you handle class imbalance?

The dataset contains more Pneumonia images than Normal images.

To address imbalance:
- Use class weights during training.
- Monitor Recall and F1 score in addition to Accuracy.
- Apply augmentation to improve generalization.

Recall is especially important because false negatives are
dangerous in medical diagnosis.
"""
```

```python
# ==========================================
# SECTION 4 - MODEL ARCHITECTURE
# ==========================================

from tensorflow import keras

# ------------------------------------------
# Transfer Learning Backbone
# ------------------------------------------

base = keras.applications.MobileNetV2(
    input_shape=(150, 150, 3),
    include_top=False,
    weights="imagenet"
)

# Freeze pretrained weights
base.trainable = False

# ------------------------------------------
# Classification Head
# ------------------------------------------

model = keras.Sequential([

    # Feature extractor
    base,

    # Convert feature maps to vector
    keras.layers.GlobalAveragePooling2D(),

    # Stabilize activations
    keras.layers.BatchNormalization(),

    # Regularization
    keras.layers.Dropout(0.30),

    # Dense classifier
    keras.layers.Dense(
        128,
        activation="relu"
    ),

    keras.layers.Dropout(0.20),

    # Binary output
    keras.layers.Dense(
        1,
        activation="sigmoid"
    )
])

model.summary()

print("\nTotal Parameters:")
print(f"{model.count_params():,}")
```

```python
# ==========================================
# SECTION 5 - TRAINING
# ==========================================

from tensorflow import keras
from tensorflow.keras.callbacks import (
    EarlyStopping,
    ModelCheckpoint,
    ReduceLROnPlateau,
    CSVLogger
)

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ==========================================
# Callbacks
# ==========================================

callbacks = [

    EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True,
        verbose=1
    ),

    ModelCheckpoint(
        "best_model.keras",
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    ),

    ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.5,
        patience=4,
        min_lr=1e-7,
        verbose=1
    ),

    CSVLogger("training_log.csv")
]

# ==========================================
# Compile Model
# ==========================================

model.compile(

    optimizer=keras.optimizers.Adam(
        learning_rate=0.0001
    ),

    loss="binary_crossentropy",

    metrics=[
        "accuracy",
        keras.metrics.AUC(name="auc"),
        keras.metrics.Recall(name="recall"),
        keras.metrics.Precision(name="precision")
    ]
)

# ==========================================
# Train
# ==========================================

history = model.fit(

    train_ds,

    validation_data=val_ds,

    epochs=50,

    callbacks=callbacks,

    verbose=1
)

# ==========================================
# Training Summary
# ==========================================

stopped_epoch = len(history.history["loss"])

best_val_loss = min(
    history.history["val_loss"]
)

best_val_acc = max(
    history.history["val_accuracy"]
)

print("\nTRAINING SUMMARY")
print("=" * 50)

print(
    f"Training stopped at epoch: {stopped_epoch}"
)

print(
    f"Best validation loss: {best_val_loss:.4f}"
)

print(
    f"Best validation accuracy: {best_val_acc:.4f}"
)

# ==========================================
# Load Training Log
# ==========================================

log_df = pd.read_csv(
    "training_log.csv"
)

print("\nTraining Log Columns:")
print(log_df.columns.tolist())

# ==========================================
# Learning Rate Extraction
# ==========================================

if "learning_rate" in log_df.columns:
    lr_values = log_df["learning_rate"]
else:
    lr_values = [0.0001] * len(log_df)

# ==========================================
# Detect LR Reductions
# ==========================================

lr_change_epochs = []

for i in range(1, len(lr_values)):

    if lr_values[i] < lr_values[i - 1]:

        lr_change_epochs.append(i)

print("\nLR Reduction Epochs:")
print(lr_change_epochs)

# ==========================================
# Training Curves
# ==========================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(18, 5)
)

# ------------------------------------------
# Loss
# ------------------------------------------

axes[0].plot(
    history.history["loss"],
    label="Train Loss"
)

axes[0].plot(
    history.history["val_loss"],
    label="Val Loss"
)

axes[0].axvline(
    stopped_epoch - 1,
    linestyle="--"
)

for epoch in lr_change_epochs:

    axes[0].axvline(
        epoch,
        linestyle=":"
    )

axes[0].set_title("Loss")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Loss")
axes[0].legend()

# ------------------------------------------
# Accuracy
# ------------------------------------------

axes[1].plot(
    history.history["accuracy"],
    label="Train Accuracy"
)

axes[1].plot(
    history.history["val_accuracy"],
    label="Val Accuracy"
)

axes[1].axvline(
    stopped_epoch - 1,
    linestyle="--"
)

for epoch in lr_change_epochs:

    axes[1].axvline(
        epoch,
        linestyle=":"
    )

axes[1].set_title("Accuracy")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Accuracy")
axes[1].legend()

# ------------------------------------------
# Learning Rate
# ------------------------------------------

axes[2].plot(
    lr_values,
    linewidth=2
)

axes[2].axvline(
    stopped_epoch - 1,
    linestyle="--"
)

for epoch in lr_change_epochs:

    axes[2].axvline(
        epoch,
        linestyle=":"
    )

axes[2].set_title("Learning Rate")
axes[2].set_xlabel("Epoch")
axes[2].set_ylabel("LR")

plt.tight_layout()

plt.savefig(
    "training_curves.png",
    dpi=150
)

plt.show()

# ==========================================
# Callback Analysis
# ==========================================

if len(lr_change_epochs) > 0:

    print(
        "\nReduceLROnPlateau Triggered"
    )

    print(
        f"Epochs: {lr_change_epochs}"
    )

else:

    print(
        "\nReduceLROnPlateau never triggered."
    )

# ==========================================
# REPORT COMMENTS
# ==========================================

"""
POST-TRAINING ANALYSIS

1. Is the model overfitting, underfitting, or well-fit?

Look at the train and validation curves.

Overfitting:
- Train loss decreases continuously.
- Validation loss starts increasing.
- Large gap between train and validation accuracy.

Underfitting:
- Both train and validation accuracy remain low.
- Both losses remain high.

Well-fit:
- Train and validation curves are close.
- Validation loss decreases steadily.
- Validation accuracy stabilizes at a high value.

------------------------------------------------

2. Did ReduceLROnPlateau help?

Check whether validation loss improves shortly
after each learning-rate reduction.

If validation loss decreases after LR reduction,
the callback successfully helped optimization
continue.

------------------------------------------------

3. Would more epochs help?

If EarlyStopping stopped training and validation
loss had already plateaued, additional epochs
are unlikely to help.

If validation loss is still decreasing at the
final epoch, more training may improve results.

The training curves should be used to justify
the answer.
"""
```

```python
# ==========================================
# SECTION 6 - MODEL EVALUATION
# ==========================================

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    roc_curve,
    roc_auc_score,
    precision_score,
    recall_score,
    f1_score,
    accuracy_score
)

# ==========================================
# Generate Predictions
# ==========================================

y_true = []
y_pred_proba = []

for images, labels in test_ds:

    preds = model.predict(
        images,
        verbose=0
    )

    y_true.extend(
        labels.numpy().flatten()
    )

    y_pred_proba.extend(
        preds.flatten()
    )

y_true = np.array(y_true)
y_pred_proba = np.array(y_pred_proba)

# Binary threshold
y_pred = (
    y_pred_proba > 0.5
).astype(int)

# ==========================================
# Confusion Matrix
# ==========================================

cm = confusion_matrix(
    y_true,
    y_pred
)

plt.figure(figsize=(7, 6))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Greens",
    xticklabels=["Normal", "Pneumonia"],
    yticklabels=["Normal", "Pneumonia"]
)

plt.title("Confusion Matrix - Test Set")
plt.xlabel("Predicted")
plt.ylabel("Actual")

plt.tight_layout()

plt.savefig(
    "confusion_matrix.png",
    dpi=150
)

plt.show()

# ==========================================
# Classification Report
# ==========================================

print("\nCLASSIFICATION REPORT")
print("=" * 50)

report = classification_report(
    y_true,
    y_pred,
    target_names=[
        "Normal",
        "Pneumonia"
    ]
)

print(report)

# ==========================================
# ROC Curve
# ==========================================

fpr, tpr, thresholds = roc_curve(
    y_true,
    y_pred_proba
)

auc = roc_auc_score(
    y_true,
    y_pred_proba
)

plt.figure(figsize=(7, 6))

plt.plot(
    fpr,
    tpr,
    linewidth=2,
    label=f"AUC = {auc:.3f}"
)

plt.plot(
    [0, 1],
    [0, 1],
    "k--"
)

plt.fill_between(
    fpr,
    tpr,
    alpha=0.1
)

plt.title("ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")

plt.legend()

plt.tight_layout()

plt.savefig(
    "roc_curve.png",
    dpi=150
)

plt.show()

# ==========================================
# Metrics
# ==========================================

accuracy = accuracy_score(
    y_true,
    y_pred
)

precision = precision_score(
    y_true,
    y_pred
)

recall = recall_score(
    y_true,
    y_pred
)

f1 = f1_score(
    y_true,
    y_pred
)

tn, fp, fn, tp = cm.ravel()

sensitivity = tp / (tp + fn)

specificity = tn / (tn + fp)

# ==========================================
# Print Metrics
# ==========================================

print("\nTEST METRICS")
print("=" * 50)

print(
    f"Accuracy     : {accuracy:.4f}"
)

print(
    f"Precision    : {precision:.4f}"
)

print(
    f"Recall       : {recall:.4f}"
)

print(
    f"F1 Score     : {f1:.4f}"
)

print(
    f"Sensitivity  : {sensitivity:.2%}"
)

print(
    f"Specificity  : {specificity:.2%}"
)

print(
    f"AUC          : {auc:.4f}"
)

# ==========================================
# False Negatives
# ==========================================

print("\nERROR ANALYSIS")
print("=" * 50)

print(f"True Negatives : {tn}")
print(f"False Positives: {fp}")
print(f"False Negatives: {fn}")
print(f"True Positives : {tp}")

# ==========================================
# Evaluation Summary
# ==========================================

"""
EVALUATION SUMMARY

1. Overall Performance

The model achieved strong performance on the chest X-ray
classification task, as measured by accuracy, precision,
recall, F1-score, and AUC. The ROC curve provides a visual
representation of the tradeoff between sensitivity and
specificity across different classification thresholds.

------------------------------------------------

2. Where Does the Model Fail Most?

The most important failures are False Negatives.

False Negative:

    Actual    = Pneumonia
    Predicted = Normal

In this situation, a patient with pneumonia would be
incorrectly classified as healthy.

------------------------------------------------

3. Is This Performance Good Enough?

For medical applications, sensitivity (recall) is often
more important than accuracy.

A model with:

    Sensitivity > 90%

may be useful as a clinical screening tool.

A model with low sensitivity would not be acceptable
because too many pneumonia cases would be missed.

------------------------------------------------

4. What Would I Improve Next?

Potential improvements:

- Fine-tune MobileNetV2 layers.
- Increase image resolution.
- Use class weights.
- Add more augmentation.
- Use ensemble models.
- Tune decision threshold.
- Collect additional training data.

------------------------------------------------

5. Medical Interpretation

A false negative means the model predicts that a patient
is healthy when the patient actually has pneumonia.

This is the most dangerous error because treatment may
be delayed or missed entirely.

For this reason, sensitivity should be prioritized over
raw accuracy when evaluating medical diagnostic systems.
"""
```

```python
# ==========================================
# SECTION 10 - PREDICTION INTERFACE
# ==========================================

import json
import logging
import pathlib
import numpy as np
import matplotlib.pyplot as plt

from datetime import datetime
from tensorflow import keras

logging.basicConfig(level=logging.INFO)

logger = logging.getLogger("DLPredictor")

# ==========================================
# Predictor Class
# ==========================================

class DeepLearningPredictor:

    def __init__(
        self,
        model_path,
        metadata_path,
        labels_path
    ):

        self.model = keras.models.load_model(
            model_path
        )

        with open(metadata_path) as f:
            self.meta = json.load(f)

        with open(labels_path) as f:
            self.labels = json.load(f)

        self.input_shape = tuple(
            self.meta["input_shape"]
        )

        logger.info(
            f"Loaded model. Test AUC = "
            f"{self.meta['metrics']['test_auc']}"
        )

    def preprocess_image(
        self,
        image_path
    ):

        img = keras.utils.load_img(
            image_path,
            target_size=self.input_shape[:2]
        )

        arr = keras.utils.img_to_array(
            img
        )

        arr = arr / 255.0

        return np.expand_dims(
            arr,
            axis=0
        )

    def predict(
        self,
        input_data
    ):

        try:

            if isinstance(
                input_data,
                str
            ):

                x = self.preprocess_image(
                    input_data
                )

            else:

                x = np.expand_dims(
                    input_data,
                    axis=0
                )

            probs = self.model.predict(
                x,
                verbose=0
            )[0]

            # Binary Classification

            pred_idx = int(
                probs[0] > 0.5
            )

            confidence = float(
                probs[0]
            )

            label = self.labels[
                str(pred_idx)
            ]

            result = {

                "prediction":
                    label,

                "confidence":
                    f"{confidence:.1%}",

                "timestamp":
                    datetime.now().isoformat()
            }

            logger.info(
                f"Prediction: {label}"
            )

            return result

        except Exception as e:

            logger.error(
                f"Prediction Error: {e}"
            )

            return {
                "error": str(e)
            }

    def predict_batch(
        self,
        inputs
    ):

        return [
            self.predict(x)
            for x in inputs
        ]


# ==========================================
# Create Predictor
# ==========================================

predictor = DeepLearningPredictor(

    "final_model.keras",

    "model_metadata.json",

    "class_labels.json"
)

# ==========================================
# Select Test Images
# ==========================================

data_dir = pathlib.Path(
    "/content/drive/MyDrive/chest_xray/chest_xray/test"
)

normal_imgs = list(
    (data_dir/"NORMAL").glob("*.jpeg")
)[:5]

pneumonia_imgs = list(
    (data_dir/"PNEUMONIA").glob("*.jpeg")
)[:5]

test_images = (
    normal_imgs +
    pneumonia_imgs
)

true_labels = (
    ["NORMAL"] * 5 +
    ["PNEUMONIA"] * 5
)

# ==========================================
# Predictions
# ==========================================

correct = 0

results = []

for path, truth in zip(
    test_images,
    true_labels
):

    pred = predictor.predict(
        str(path)
    )

    predicted = pred["prediction"]

    is_correct = (
        predicted == truth
    )

    if is_correct:
        correct += 1

    print(
        f"\nFile: {path.name}"
    )

    print(
        f"True Label : {truth}"
    )

    print(
        f"Predicted  : {predicted}"
    )

    print(
        f"Confidence : {pred['confidence']}"
    )

    print(
        f"Correct    : {is_correct}"
    )

    results.append(
        (
            path,
            truth,
            pred,
            is_correct
        )
    )

if len(test_images) == 0:
    raise ValueError(
        "No test images found. Check data_dir."
    )

accuracy = correct / len(
    test_images
)

print(
    f"\n10-Sample Accuracy: "
    f"{accuracy:.2%}"
)

# ==========================================
# Prediction Grid
# ==========================================

fig, axes = plt.subplots(
    2,
    5,
    figsize=(15, 6)
)

axes = axes.flatten()

for ax, (
    path,
    truth,
    pred,
    is_correct
) in zip(axes, results):

    img = keras.utils.load_img(
        path,
        target_size=(150, 150)
    )

    ax.imshow(img)

    ax.set_title(
        f"{pred['prediction']}\n"
        f"{pred['confidence']}"
    )

    color = (
        "green"
        if is_correct
        else "red"
    )

    for spine in ax.spines.values():

        spine.set_edgecolor(
            color
        )

        spine.set_linewidth(
            4
        )

    ax.axis("off")

plt.tight_layout()

plt.savefig(
    "prediction_grid.png",
    dpi=150
)

plt.show()

# ==========================================
# Error Handling Tests
# ==========================================

print("\nERROR TEST 1")
print("=" * 40)

bad_path = predictor.predict(
    "does_not_exist.jpeg"
)

print(bad_path)

print("\nERROR TEST 2")
print("=" * 40)

try:

    wrong_input = np.random.rand(
        10,
        10
    )

    result = predictor.predict(
        wrong_input
    )

    print(result)

except Exception as e:

    print(
        "Handled:",
        e
    )

# ==========================================
# REPORT COMMENTS
# ==========================================

"""
SECTION 10 SUMMARY

The DeepLearningPredictor class provides a reusable
deployment interface that separates model loading,
preprocessing, and prediction logic from the training code.

The predictor automatically loads metadata and class
labels so predictions can be interpreted correctly.

The interface supports image-file prediction and batch
prediction while handling common runtime errors safely.

A prediction grid was generated to visually inspect
model behaviour on unseen test samples.

Error handling was verified by testing invalid file
paths and malformed inputs, ensuring graceful failure
instead of application crashes.
"""
```
